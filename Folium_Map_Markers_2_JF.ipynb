### Task 2: Mark the success/failed launches for each site on the map

Add the launch outcomes for each site, determine which sites have high success rates. The data frame spacex_df class column indicates if a launch was successful or not

```python
spacex_df.tail(10)
```

Note:  A launch only happens in one of four launch sites. Many launch records will have the same coordinate. Use `MarkerCluster` to simplify a map containing many markers having the same coordinates.

```python
marker_cluster = MarkerCluster()
```

```python
# Create a new column in 'spacex_df' dataframe called 'marker_color'
# Store the marker colors based on the 'class' value

# Class = 1 = Success = Green
# Class = 0 = Fail = Red

# Function to assign color to launch outcome
def mark_color(launch):
    if launch == 1:
        return 'green'
    else:
        return 'red'
    
spacex_df['marker_color'] = spacex_df['class'].apply(mark_color)
spacex_df.tail(10)
```

```python
spacex_df.tail(10)
```

```python
site_map.add_child(marker_cluster)

spacex_df['map_coord'] = spacex_df.apply(lambda x: [x['Lat'], x['Long']], axis = 1)

for index, record in spacex_df.iterrows():
    marker = folium.Marker(
        location = spacex_df['map_coord'][index],
        icon = folium.Icon(color='white'),
        icon_color = spacex_df['marker_color'][index]).add_to(marker_cluster)
    
marker_cluster.add_child(marker)

site_map
```

### TASK 3: Calculate the distances between a launch site to its proximities

```python


from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```
